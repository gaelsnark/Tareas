**R-12.7** Suppose we are given two $n$-element sorted sequences $A$ and $B$ each with distinct elements, but potentially some elements that are in both sequences. Implement an $O(n)$-time method for computing a sequence representing the union $A \cup B$ (with no duplicates) as a sorted sequence.

```python
def union(A, B):
    i = j = 0
    R = []
    while i < len(A) or j < len(B):
        if j==len(B) or (i < len(A) and A[i] < B[j]):
            R.append(A[i])      # toma de A
            i += 1
        elif i == len(A) or A[i] > B[j]:
            R.append(B[j])      
            j += 1
        else:                   
            R.append(A[i])
            i += 1
            j += 1
    return R

print(union([8,1,2,5,4], [0,12,3,1,2,8]))

```

En cada vuelta avanza i o j o ambos, nunca retroceden, así que hay como máximo n + n = 2n vueltas, cada una con trabajo O(1)


**R-12.11** Suppose function inplace_quick_sort is executed on a sequence with duplicate elements. Prove that the algorithm still correctly sorts the input sequence. What happens in the partition step when there are elements equal to the pivot? What is the running time of the algorithm if all the input elements are equal?

```python
def inpl_ace_quicksort_(S, a, b):
    if a >= b: return
    pivot = S[b]
    left, right = a, b - 1
    while left <= right:
        while left <= right and S[left] < pivot:  left += 1
        while left <= right and pivot < S[right]: right -= 1
        if left <= right:
            S[left], S[right] = S[right], S[left]
            left, right = left + 1, right - 1
    S[left], S[b] = S[b], S[left]     
    
     
    inpl_ace_quicksort_(S, a, left - 1)
    inpl_ace_quicksort_(S, left + 1, b)
```

left avanza solo mientras S[left]< pivote. Si encuentra un elemento igual al pivote, la condición S[left]<pivote es falsa y se detiene.
right retrocede solo mientras pivote <S[right]. Si encuentra un elemento igual al pivote, la condición es falsa y se detiene.

Ninguno de los dos se pasa de largo sobre un elemento igual al pivote, Los dos se paran ahí.

**R-12.21** Given a sequence $S$ of $n$ values, each equal to 0 or 1, describe an in-place method for sorting $S$.

```python
def sort_01(S):
    left, right = 0, len(S) - 1
    
    while left < right:
        if S[left] == 0:        
            left += 1
        elif S[right] == 1:        
            right -= 1
        else:                      
            S[left], S[right] = S[right], S[left]
            left += 1
            right -= 1

S = [0,1,0,1,1,1,0,1,1]
sort_01(S)
print(S)
```

**C-12.25** Linda claims to have an algorithm that takes an input sequence $S$ and produces an output sequence $T$ that is a sorting of the $n$ elements in $S$.

a. Give an algorithm, is_sorted, that tests in $O(n)$ time if $T$ is sorted.

b. Explain why the algorithm is_sorted is not sufficient to prove a particular output $T$ to Linda’s algorithm is a sorting of $S$.

c. Describe what additional information Linda’s algorithm could output so that her algorithm’s correctness could be established on any given $S$ and $T$ in $O(n)$ time.

```python
def Ais_sorted(T):
    for i in range(len(T) - 1):
        if T[i] > T[i+1]:
            return False
    return True
```

b) Se recorre T una vez comparando cada elemento con el siguiente. Son n − 1 comparaciones de O(1), así que el tiempo es O(n).
is_sorted solo dice que T está en orden, pero no dice que T tenga los mismos elementos que S.

```python
def verify(S,T,P):
    n = len(S)
    if len(T)!= n or len(P) != n:
        return False
    vistos = [False]*n
    for i in range(n):
        if not (0 <= P[i] < n) or vistos[P[i]]:
            return False          
        vistos[P[i]] = True
        if T[i]!= S[P[i]]:
            return False          
    return Ais_sorted(T) 
```

**C-12.26** Describe and analyze an efficient method for removing all duplicates from a collection $A$ of $n$ elements.

```python
def removin_duplicates(A):
    A.sort()                   
    if not A:
        return []
    R = [A[0]]
    for i in range(1, len(A)):
        if A[i] != A[i-1]:        
            R.append(A[i])
    return R

print(removin_duplicates([0,1,3,2,8,7,9]))
```

**C-12.27** Augment the PositionalList class (see Section 7.4) to support a method named merge with the following behavior. If $A$ and $B$ are PositionalList instances whose elements are sorted, the syntax A.merge(B) should merge all elements of $B$ into $A$ so that $A$ remains sorted and $B$ becomes empty. Your implementation must accomplish the merge by relinking existing nodes; you are not to create any new nodes.

```python
def merge(self,other):
    a = self._header._next             
    b = other._header._next            

    while a is not self._trailer and b is not other._trailer:
        if b._element<a._element:    
            sig = b._next
            b._prev = a._prev
            b._next = a
            a._prev._next=b
            a._prev = b
            b = sig
        else:                          
            a = a._next
            
    if b is not other._trailer:       
        ultimo_b = other._trailer._prev
        ultimo_a = self._trailer._prev
        ultimo_a._next = b
        b._prev = ultimo_a
        ultimo_b._next = self._trailer
        self._trailer._prev = ultimo_b
        
    self._size += other._size         
    
    
    other._header._next = other._trailer   
    other._trailer._prev = other._header
    other._size = 0
```

**C-12.28** Augment the PositionalList class (see Section 7.4) to support a method named sort that sorts the elements of a list by relinking existing nodes; you are not to create any new nodes. You may use your choice of sorting algorithm.

```python
def sort(self):
    if self._size < 2:
        return
    marker = self._header._next          
    while marker._next is not self._trailer:
        pivot = marker._next             
        if pivot._element >= marker._element:
            marker = pivot            
               
        else:
            marker._next = pivot._next
            pivot._next._prev = marker
            walk = marker
            
            while walk._prev is not self._header and walk._prev._element > pivot._element:
                walk = walk._prev
            pivot._prev = walk._prev
            pivot._next = walk
            walk._prev._next = pivot
            walk._prev = pivot
```
